### 3: Step 3: Dokümanın Bölümlere Ayrılması


**3A**

Dokümandaki mevcut başlıklar kullanılarak metin bölümlere ayrılacaktır.

Her bölüm daha sonra uygun büyüklükte chunk'lara bölünecektir.

Her chunk için "kaynak", "sayfa" ve "bolum" metadata alanları tutulacaktır.

"bolum" alanı, metadata filtreleme işlemlerinde kullanılacaktır.

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Dokümanda bulunan gerçek bölüm başlıkları
ana_basliklar = {
    "Hatay Handmade Hakkında",
    "Hatay Handmade'in Desteklediği Değerler",
    "Ürünler ve Üretim Alanları",
    "Üretim, Tedarik ve İşleyiş",
    "Üretim ve Tedarik Yapısı",
    "Yerel Üretim ve Geleneksel İşçilik",
    "Tedarik ve Stoklama",
    "Fiyatlandırma",
    "Paketleme ve Kargo",
    "Müşteri Memnuniyeti ve Telafi",
    "Sipariş Takibi ve Müşteri Desteği",
    "Gelecek Vizyonu"
}

def bolumleri_ayir(sayfalar):
    bolumler = []
    mevcut_bolum = "Genel"

    for sayfa in sayfalar:
        satirlar = sayfa.page_content.splitlines()
        metin = []

        for satir in satirlar:
            satir = satir.strip()

            if not satir or satir.isdigit():
                continue

            # Ana başlık kontrolü
            if satir in ana_basliklar:
                if metin:
                    bolumler.append({
                        "metin": "\n".join(metin),
                        "sayfa": sayfa.metadata["sayfa"],
                        "bolum": mevcut_bolum
                    })
                    metin = []

                mevcut_bolum = satir
                continue

            # Numaralı ürün kategorileri
            if satir[:2].rstrip(".").isdigit() and ". " in satir:
                if metin:
                    bolumler.append({
                        "metin": "\n".join(metin),
                        "sayfa": sayfa.metadata["sayfa"],
                        "bolum": mevcut_bolum
                    })
                    metin = []

                mevcut_bolum = satir
                continue

            metin.append(satir)

        if metin:
            bolumler.append({
                "metin": "\n".join(metin),
                "sayfa": sayfa.metadata["sayfa"],
                "bolum": mevcut_bolum
            })

    return bolumler


bolumler = bolumleri_ayir(sayfalar)

print("Oluşturulan bölüm sayısı:", len(bolumler))

for i, bolum in enumerate(bolumler[:15], start=1):
    print(f"{i}. {bolum['bolum']} | Sayfa: {bolum['sayfa']}")
```

### 3B: Chunking İşlemi


**3BA**

Doküman bölümleri `RecursiveCharacterTextSplitter` kullanılarak chunk'lara ayrılacaktır.

Chunk boyutu 500 karakter, overlap değeri 75 karakter olarak kullanılacaktır.

Her chunk için kaynak, sayfa ve bölüm bilgileri metadata olarak korunacaktır.

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=75
)

chunklar = []

for bolum in bolumler:
    dokuman = Document(
        page_content=bolum["metin"],
        metadata={
            "kaynak": PDF_YOLU,
            "sayfa": bolum["sayfa"],
            "bolum": bolum["bolum"]
        }
    )

    bolum_chunklari = splitter.split_documents([dokuman])
    chunklar.extend(bolum_chunklari)

print("Toplam chunk sayısı:", len(chunklar))

ortalama_uzunluk = sum(
    len(chunk.page_content) for chunk in chunklar
) / len(chunklar)

print("Ortalama chunk uzunluğu:", round(ortalama_uzunluk, 2))
print("\nÖrnek chunk metadata:")
print(chunklar[0].metadata)

print("\nÖrnek chunk:")
print(chunklar[0].page_content)
```

## 4: Step 4: Embedding ve Chroma


### 4A: Embedding Modelinin Hazırlanması


**4AA**

Doküman chunk'ları `paraphrase-multilingual-mpnet-base-v2` embedding modeli ile vektörlere dönüştürülecektir.

Bu model Türkçe metinlerle çalışmak için kullanılacaktır.

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/paraphrase-multilingual-mpnet-base-v2"
)

print("Embedding modeli hazır.")
```

### 4B: Chroma Vector Store'un Oluşturulması


**4BA**

Hazırlanan chunk'lar embedding modeli kullanılarak Chroma vector store'a aktarılacaktır.

Arama yöntemi olarak cosine similarity kullanılacaktır.

```python
from langchain_chroma import Chroma

vectorstore = Chroma.from_documents(
    documents=chunklar,
    embedding=embedding_model,
    collection_name="hatay_handmade",
    collection_metadata={"hnsw:space": "cosine"}
)

print("Chroma vector store hazır.")
print("Chunk sayısı:", vectorstore._collection.count())
```

### 4C: Filtresiz ve Metadata Filtreli Retriever


**4CA**

Aynı Chroma veritabanı üzerinden iki farklı retriever oluşturulacaktır.

İlk retriever tüm chunk'lar üzerinde arama yapacaktır.

İkinci retriever ise `bolum` metadata alanını kullanarak aramayı belirli bir bölümle sınırlandıracaktır.

```python
retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 3}
)

filtreli_retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={
        "k": 3,
        "filter": {
            "bolum": "Tedarik ve Stoklama"
        }
    }
)

print("Filtresiz retriever hazır.")
print("Metadata filtreli retriever hazır.")
```

### 4D: Retriever Sonuçlarının Karşılaştırılması


**4DA**

Aynı soru önce filtresiz retriever'a, ardından `Tedarik ve Stoklama` filtresi uygulanmış retriever'a gönderilecektir.

Her iki sonucun bölüm ve sayfa metadata bilgileri karşılaştırılacaktır.

```python
soru = "Stokta bulunmayan ürünler için ne yapılabilir?"

sonuclar_filtresiz = retriever.invoke(soru)
sonuclar_filtreli = filtreli_retriever.invoke(soru)

print("=== FİLTRESİZ RETRIEVER ===")

for i, sonuc in enumerate(sonuclar_filtresiz, start=1):
    print(f"\n{i}.")
    print("Bölüm:", sonuc.metadata["bolum"])
    print("Sayfa:", sonuc.metadata["sayfa"])
    print("Metin:", sonuc.page_content[:300])


print("\n" + "=" * 60)
print("=== METADATA FİLTRELİ RETRIEVER ===")

for i, sonuc in enumerate(sonuclar_filtreli, start=1):
    print(f"\n{i}.")
    print("Bölüm:", sonuc.metadata["bolum"])
    print("Sayfa:", sonuc.metadata["sayfa"])
    print("Metin:", sonuc.page_content[:300])
```

## 5: Step 5: Retrieval Stratejilerinin Karşılaştırılması


### 5A: Similarity, MMR ve Threshold Retriever'ların Oluşturulması


**5AA**

Aynı Chroma vector store üzerinden üç farklı retrieval stratejisi oluşturulacaktır.

Threshold retriever için başlangıç benzerlik eşiği `0.3` olarak kullanılacaktır.

```python
similarity_retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 3}
)

mmr_retriever = vectorstore.as_retriever(
    search_type="mmr",
    search_kwargs={
        "k": 3,
        "fetch_k": 6,
        "lambda_mult": 0.5
    }
)

threshold_retriever = vectorstore.as_retriever(
    search_type="similarity_score_threshold",
    search_kwargs={
        "score_threshold": 0.3,
        "k": 3
    }
)

print("Similarity retriever hazır.")
print("MMR retriever hazır.")
print("Threshold retriever hazır.")
```

### 5B: Üç Retrieval Stratejisinin Karşılaştırılması


**5BA**

Aynı soru Similarity, MMR ve Threshold retriever'larına verilecektir.

Her yöntemin döndürdüğü chunk'ların bölüm ve sayfa bilgileri karşılaştırılacaktır.

```python
soru = "Hatay Handmade stokta olmayan ürünler için ne yapabilir?"

sonuclar_similarity = similarity_retriever.invoke(soru)
sonuclar_mmr = mmr_retriever.invoke(soru)
sonuclar_threshold = threshold_retriever.invoke(soru)


def sonuclari_yazdir(baslik, sonuclar):
    print(f"\n{'=' * 60}")
    print(baslik)
    print(f"{'=' * 60}")

    if not sonuclar:
        print("Sonuç bulunamadı.")
        return

    for i, sonuc in enumerate(sonuclar, start=1):
        print(f"\n{i}. Bölüm: {sonuc.metadata['bolum']}")
        print(f"   Sayfa: {sonuc.metadata['sayfa']}")
        print(f"   Metin: {sonuc.page_content[:250]}")


sonuclari_yazdir("SIMILARITY", sonuclar_similarity)
sonuclari_yazdir("MMR", sonuclar_mmr)
sonuclari_yazdir("THRESHOLD", sonuclar_threshold)
```

## 6: Step 6: LCEL RAG Pipeline


### 6A: Retrieved Context'in Biçimlendirilmesi


**6AA**

Retriever tarafından bulunan chunk'lar tek bir context metnine dönüştürülecektir.

Her chunk'ın kaynak dosyası ve sayfa numarası context içinde korunacaktır.

```python
def bicimle(dokumanlar):
    parcalar = []

    for dokuman in dokumanlar:
        kaynak = dokuman.metadata["kaynak"]
        sayfa = dokuman.metadata["sayfa"]

        parcalar.append(
            f"[Kaynak: {kaynak} | Sayfa: {sayfa}]\n"
            f"{dokuman.page_content}"
        )

    return "\n\n".join(parcalar)


print("Context biçimlendirme fonksiyonu hazır.")
```

### 6B: RAG Sistem Talimatının Tanımlanması


**6BA**

RAG modeli yalnızca kendisine verilen doküman context'ini kullanacaktır.

Context içinde bulunmayan bilgiler için `Bu bilgi dokümanda yok.` ifadesi kullanılacaktır.

Cevabın sonunda kullanılan kaynak dosyası ve sayfa bilgisi belirtilecektir.

```python
RAG_SISTEM = """
Sen Hatay Handmade doküman asistanısın.

Yalnızca verilen CONTEXT içindeki bilgilere dayanarak cevap ver.

Sorunun cevabı CONTEXT içinde yoksa:
"Bu bilgi dokümanda yok." de.

Context dışında bilgi ekleme ve tahmin yapma.

Cevabın sonunda kullanılan kaynak dosyası ve sayfa numarasını belirt.
"""

print("RAG sistem talimatı hazır.")
```

### 6CB: Context Zincirinin Tanımlanması

**6CBA**

Retriever ve `bicimle` fonksiyonu birleştirilerek RAG zincirinin kullanacağı context zinciri hazırlanacaktır.

```python
from langchain_core.runnables import RunnableLambda

context_zinciri = (
    retriever
    | RunnableLambda(bicimle)
)

print("Context zinciri hazır.")
```

### 6C: LCEL RAG Zincirinin Oluşturulması


**6CA**

RAG akışı LCEL kullanılarak oluşturulacaktır.

Akış; retriever, context biçimlendirme, soru ve context'in birleştirilmesi ve `sor()` fonksiyonuyla cevap üretme adımlarından oluşacaktır.

```python
from langchain_core.runnables import (
    RunnableLambda,
    RunnableParallel,
    RunnablePassthrough
)

rag_zinciri = (
    RunnableParallel(
        context=context_zinciri,
        soru=RunnablePassthrough()
    )
    | RunnableLambda(
        lambda x : sor(
            f"""
{RAG_SISTEM}

CONTEXT :
{x["context"]}

KULLANICI SORUSU :
{x["soru"]}
"""
        )
    )
)

print("LCEL RAG zinciri hazır.")
```

### 6D: İlk RAG Testi


**6DA**

LCEL RAG zinciri Hatay Handmade dokümanında bulunan bir soru ile test edilecektir.

Cevabın yalnızca retrieved context'e dayanması ve kaynak bilgisini içermesi beklenmektedir.

```python
soru = "Hatay Handmade stokta bulunmayan ürünler için ne yapabilir?"

cevap = rag_zinciri.invoke(soru)

print("SORU:")
print(soru)

print("\nCEVAP:")
print(cevap)
```

### 6E: Retrieved Context'in Kontrol Edilmesi


**6EA**

RAG zincirinin oluşturduğu context doğrudan görüntülenecektir.

Böylece modelin cevabında kullanılabilecek doküman parçalarının tamamı kontrol edilecektir.

```python
soru = "Hatay Handmade stokta bulunmayan ürünler için ne yapabilir?"

dokumanlar = retriever.invoke(soru)
context = bicimle(dokumanlar)

print(context)
```

**6FA**

Sorunun cevabı aynı bölümdeki bir sonraki chunk'ta bulunduğu için RAG zincirinde daha fazla aday chunk getirilecektir.

```python
rag_retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 5}
)

print("RAG retriever hazır.")
print("Toplam getirilecek chunk:", 5)
```

### 6G: Genişletilmiş Retriever'ın Testi


**6GA**

Retriever `k=5` olacak şekilde genişletilmiştir.

Aynı soru tekrar aranacaktır.

Amaç, “Stok geldiğinde haber ver” bilgisinin retrieved context içinde bulunup bulunmadığını kontrol etmektir.

```python
soru = "Hatay Handmade stokta bulunmayan ürünler için ne yapabilir?"

dokumanlar = rag_retriever.invoke(soru)
context = bicimle(dokumanlar)

print(context)
```

### 6H: Aranan Bilginin Chunk İçinde Bulunması


**6HA**

“Stok geldiğinde haber ver” ifadesinin hangi chunk içinde bulunduğu doğrudan kontrol edilecektir.

Böylece bilginin chunk'lara doğru aktarılıp aktarılmadığı doğrulanacaktır.

### 7A: Doküman Arama Aracının Tanımlanması

**7AA**

`dokumanda_ara` aracı, `Tedarik ve Stoklama` bölümündeki en ilgili üç parçayı bulacak ve kaynak sayfasıyla birlikte döndürecektir.

```python
from langchain_core.tools import tool

@tool
def dokumanda_ara(soru: str) -> str:
    """Tedarik ve Stoklama bölümünde soruyla ilgili en fazla 3 doküman parçasını bulur."""
    dokumanlar = filtreli_retriever.invoke(soru)

    if not dokumanlar:
        return "Tedarik ve Stoklama bölümünde ilgili bilgi bulunamadı."

    return "\n\n".join(
        f"[Kaynak: {dokuman.metadata['kaynak']} | "
        f"Bölüm: {dokuman.metadata['bolum']} | "
        f"Sayfa: {dokuman.metadata['sayfa']}]\n"
        f"{dokuman.page_content}"
        for dokuman in dokumanlar[:3]
    )

print("Araç adı:", dokumanda_ara.name)
print("Parametreler:", dokumanda_ara.args)
print("Açıklama:", dokumanda_ara.description)
```

### 7C: CSV Ürün Sorgulama Aracının Tanımlanması

**7CA**

`csv_urun_bul` aracı, ürün adını `urunler_df` içinde tam eşleşmeyle arayacaktır. Ürün bulunamazsa açıklayıcı bir mesaj döndürecektir.

```python
from langchain_core.tools import tool

@tool
def csv_urun_bul(urun_adi: str) -> str:
    """Ürün adını CSV verisinde tam eşleşmeyle arar."""
    aranan = urun_adi.strip().casefold()

    eslesenler = urunler_df.loc[
        urunler_df["urun_adi"]
        .astype(str)
        .str.strip()
        .str.casefold()
        == aranan
    ]

    if eslesenler.empty:
        return f"'{urun_adi}' adıyla eşleşen ürün bulunamadı."

    alanlar = [
        "urun_adi",
        "sku",
        "kategori",
        "fiyat",
        "stok",
        "stok_tipi",
        "satisa_acik",
        "varyantlar"
    ]

    return "\n".join(
        " | ".join(f"{alan}: {kayit[alan]}" for alan in alanlar)
        for _, kayit in eslesenler.iterrows()
    )

print("Araç adı:", csv_urun_bul.name)
print("Parametreler:", csv_urun_bul.args)
print("Açıklama:", csv_urun_bul.description)
```

### 7D: CSV Ürün Arama Aracının Elle Test Edilmesi

**7DA**

Araç önce CSV'de bulunan `Nar ekşisi` ürünüyle, sonra bulunmayan bir ürün adıyla çağrılacaktır. Tam eşleşme ve bulunamadı mesajı kontrol edilecektir.

```python
urun_sonucu = csv_urun_bul.invoke({
    "urun_adi": "Nar ekşisi"
})

bulunamayan_urun_sonucu = csv_urun_bul.invoke({
    "urun_adi": "Hayali ürün"
})

print("BULUNAN ÜRÜN TESTİ:")
print(urun_sonucu)

print("\nÜRÜN BULUNAMADI TESTİ:")
print(bulunamayan_urun_sonucu)
```
